## create a pandas dataframe from pandas
we can do or full analysis using numpy or numpy array but we will create a pandas Dataframe because it does make something simpler and also to get some practice of how to use pandas

```python
df = pd.DataFrame(data["data"],columns=data["feature_names"])
```

```python
df["target"] = data["target"]
```

```python
print(df.head())
```

```python
df.describe()
```

## Distribution of feature and target

```python
df["sepal length (cm)"].hist() ## used to plot histogram of sepal length syntax is df["column name"].hist()
plt.suptitle("sepal length histogram") # used to show the title
plt.show()
```

```python
df["sepal width (cm)"].hist() ## used to plot histogram of sepal width syntax is df["column name"].hist()
plt.suptitle("sepal width histogram") # used to show the title
plt.show()
```

```python
df["petal length (cm)"].hist() ## used to plot histogram of petal length syntax is df["column name"].hist()
plt.suptitle("petal length histogram") # used to show the title
plt.show()
```

```python
df["petal width (cm)"].hist() ## used to plot histogram of petal width syntax is df["column name"].hist() 
# df["petal width (cm)"].hist(bins=12)  used to plot the histrogram with 12 bins syntax is df["column name"].hist(bins=number of bins)
plt.suptitle("petal width histogram") # used to show the title
plt.show()
```

## reationship of the data feature with the target

```python
sns.relplot( x="petal width (cm)", y="target", hue="target", data=df) ## used to plot the scatter plot 
## syntax is sns.relplot(x="column name for x axis", y="column name for y axis", hue="column name for color coding", data=dataframe name)
```

```python
data["target_names"]
```

```python
df["target_name"] = df["target"].map({0:'setosa',1:'versicolor',2:'virginica'}) # used to map the target  with target name
```

```python
df
```

```python
sns.relplot( x="petal width (cm)", y="target", hue="target_name", data=df)
_= plt.suptitle("scatter plot of petal width vs target with target name as color coding",y = 1.05) ## used to plot the scatter plot with target name as color coding
```

```python
sns.relplot( x="petal length (cm)", y="target", hue="target_name", data=df)
_= plt.suptitle("scatter plot of petal length vs target with target name as color coding",y = 1.05) ## used to plot the scatter plot with target name as color coding
```

```python
sns.relplot( x="sepal width (cm)", y="target", hue="target_name", data=df)
_= plt.suptitle("scatter plot of sepal width vs target with target name as color coding",y = 1.05) ## used to plot the scatter plot with target name as color coding
```

```python
sns.relplot( x="sepal length (cm)", y="target", hue="target_name", data=df)
_= plt.suptitle("scatter plot of sepal length vs target with target name as color coding",y = 1.05) ## used to plot the scatter plot with target name as color coding
```
